# 01 – Data Inspection

## Project: Decathlon Retail Performance Analytics

### Objective
Inspect the raw Sales, Products, and Store datasets to understand their structure, columns, data types, record counts, and data quality before performing any cleaning or transformation.

### Datasets
1. Sales – Transaction-level retail sales data.
2. Products – Product details and classifications.
3. Store – Store details and geographical information.

### Inspection Scope
- Dataset dimensions and column names
- Data types
- Missing values
- Duplicate records
- Unique values and key columns
- Date coverage
- Relationships between datasets

```python
# Import required libraries
import pandas as pd
import numpy as np
from pathlib import Path

# Locate the project root automatically
current_path = Path.cwd()

project_root = next(
    (path for path in [current_path, *current_path.parents]
     if (path / "Data" / "Raw").exists()),
    None
)

if project_root is None:
    raise FileNotFoundError("Project root or Data/Raw folder not found.")

# Define raw data folder
raw_data_path = project_root / "Data" / "Raw"

print("Project Root:", project_root)
print("Raw Data Folder:", raw_data_path)

# Verify source files
required_files = ["sales.xlsx", "products.xlsx", "store.xlsx"]

for file in required_files:
    file_path = raw_data_path / file

    if file_path.exists():
        print(f"FOUND: {file} | Size: {file_path.stat().st_size / 1024:.2f} KB")
    else:
        print(f"MISSING: {file}")
```

```python
# Load the three raw datasets
sales_df = pd.read_excel(raw_data_path / "sales.xlsx")
products_df = pd.read_excel(raw_data_path / "products.xlsx")
store_df = pd.read_excel(raw_data_path / "store.xlsx")

print("All datasets loaded successfully.")
```

```python
# Check dataset dimensions

print("SALES DATASET")
print("Rows:", sales_df.shape[0])
print("Columns:", sales_df.shape[1])

print("\nPRODUCTS DATASET")
print("Rows:", products_df.shape[0])
print("Columns:", products_df.shape[1])

print("\nSTORE DATASET")
print("Rows:", store_df.shape[0])
print("Columns:", store_df.shape[1])
```

```python
# Inspect column names

print("SALES DATASET COLUMNS")
print(sales_df.columns.tolist())

print("\nPRODUCTS DATASET COLUMNS")
print(products_df.columns.tolist())

print("\nSTORE DATASET COLUMNS")
print(store_df.columns.tolist())
```

```python
# Preview first five records from each dataset

print("SALES DATASET")
display(sales_df.head())

print("PRODUCTS DATASET")
display(products_df.head())

print("STORE DATASET")
display(store_df.head())
```

```python
# Inspect data types

print("SALES DATA TYPES")
print(sales_df.dtypes)

print("\nPRODUCTS DATA TYPES")
print(products_df.dtypes)

print("\nSTORE DATA TYPES")
print(store_df.dtypes)
```

```python
# Check missing values in each dataset

print("SALES - MISSING VALUES")
print(sales_df.isnull().sum())

print("\nPRODUCTS - MISSING VALUES")
print(products_df.isnull().sum())

print("\nSTORE - MISSING VALUES")
print(store_df.isnull().sum())
```

```python
# Inspect promotion categories and frequencies

print("PROMOTION VALUE COUNTS")
print(sales_df["Promotion"].value_counts(dropna=False))

print("\nUNIQUE PROMOTION VALUES")
print(sales_df["Promotion"].unique())
```

```python
# Check complete duplicate records

print("SALES DATASET")
print("Duplicate rows:", sales_df.duplicated().sum())

print("\nPRODUCTS DATASET")
print("Duplicate rows:", products_df.duplicated().sum())

print("\nSTORE DATASET")
print("Duplicate rows:", store_df.duplicated().sum())
```

```python
# Check primary-key uniqueness

print("SALES")
print("Total records:", len(sales_df))
print("Unique Sale_Line_ID:", sales_df["Sale_Line_ID"].nunique())
print("Duplicate Sale_Line_ID:", sales_df["Sale_Line_ID"].duplicated().sum())

print("\nPRODUCTS")
print("Total records:", len(products_df))
print("Unique Product_Key:", products_df["Product_Key"].nunique())
print("Duplicate Product_Key:", products_df["Product_Key"].duplicated().sum())

print("\nSTORE")
print("Total records:", len(store_df))
print("Unique Store_Key:", store_df["Store_Key"].nunique())
print("Duplicate Store_Key:", store_df["Store_Key"].duplicated().sum())
```

```python
# Inspect order-level structure

print("Total sales lines:", len(sales_df))
print("Unique orders:", sales_df["Order_ID"].nunique())

print(
    "Sales lines belonging to repeated orders:",
    sales_df["Order_ID"].duplicated().sum()
)

print("\nOrders with multiple sales lines:")
order_line_counts = sales_df["Order_ID"].value_counts()

print(order_line_counts[order_line_counts > 1].value_counts().sort_index())
```

```python
# Check whether orders span multiple stores or channels

order_consistency = sales_df.groupby("Order_ID").agg(
    Store_Count=("Store_ID", "nunique"),
    Channel_Count=("Channel", "nunique")
)

print("Orders associated with multiple stores:")
print((order_consistency["Store_Count"] > 1).sum())

print("\nOrders associated with multiple channels:")
print((order_consistency["Channel_Count"] > 1).sum())
```

```python
# Inspect purchase date coverage

print("Earliest purchase date:", sales_df["Purchase_Date"].min())
print("Latest purchase date:", sales_df["Purchase_Date"].max())

print("Distinct purchase dates:", sales_df["Purchase_Date"].nunique())

print("\nRecords by year:")
print(sales_df["Purchase_Date"].dt.year.value_counts().sort_index())
```

```python
# Inspect sales records by month

monthly_records = (
    sales_df.groupby(sales_df["Purchase_Date"].dt.to_period("M"))
    .size()
    .rename("Sales_Records")
)

print(monthly_records)
```

```python
# Validate foreign-key relationships

valid_product_ids = products_df["Product_Key"]
valid_store_ids = store_df["Store_Key"]

unmatched_products = sales_df[
    ~sales_df["Product_ID"].isin(valid_product_ids)
]

unmatched_stores = sales_df[
    ~sales_df["Store_ID"].isin(valid_store_ids)
]

print("PRODUCT RELATIONSHIP")
print("Unmatched Product_ID records:", len(unmatched_products))
print("Matched Product_ID records:", len(sales_df) - len(unmatched_products))

print("\nSTORE RELATIONSHIP")
print("Unmatched Store_ID records:", len(unmatched_stores))
print("Matched Store_ID records:", len(sales_df) - len(unmatched_stores))
```

```python
# Inspect numerical columns in Sales

sales_numeric = sales_df.select_dtypes(include="number")

display(sales_numeric.describe().T)
```

```python
# Validate business ranges

print("QUANTITY VALIDATION")
print("Quantity <= 0:", (sales_df["Quantity"] <= 0).sum())

print("\nRETURN QUANTITY VALIDATION")
print("Returned_Qty < 0:", (sales_df["Returned_Qty"] < 0).sum())
print(
    "Returned_Qty > Quantity:",
    (sales_df["Returned_Qty"] > sales_df["Quantity"]).sum()
)

print("\nORDER HOUR VALIDATION")
print("Order_Hour < 0:", (sales_df["Order_Hour"] < 0).sum())
print("Order_Hour > 23:", (sales_df["Order_Hour"] > 23).sum())

print("\nDISCOUNT VALIDATION")
print("Discount_Pct < 0:", (sales_df["Discount_Pct"] < 0).sum())
print("Discount_Pct > 1:", (sales_df["Discount_Pct"] > 1).sum())
```

```python
# Validate financial calculations

tolerance = 0.01

gross_sales_diff = (
    sales_df["Gross_Sales_INR"]
    - sales_df["Quantity"] * sales_df["Unit_List_Price_INR"]
)

sales_before_returns_diff = (
    sales_df["Sales_Before_Returns_INR"]
    - (sales_df["Gross_Sales_INR"] - sales_df["Discount_INR"])
)

net_sales_diff = (
    sales_df["Net_Sales_INR"]
    - (sales_df["Sales_Before_Returns_INR"] - sales_df["Refund_INR"])
)

gross_profit_diff = (
    sales_df["Gross_Profit_INR"]
    - (sales_df["Net_Sales_INR"] - sales_df["Net_COGS_INR"])
)

print("GROSS SALES MISMATCHES:", (gross_sales_diff.abs() > tolerance).sum())

print(
    "SALES BEFORE RETURNS MISMATCHES:",
    (sales_before_returns_diff.abs() > tolerance).sum()
)

print("NET SALES MISMATCHES:", (net_sales_diff.abs() > tolerance).sum())

print("GROSS PROFIT MISMATCHES:", (gross_profit_diff.abs() > tolerance).sum())
```

```python
# Investigate zero net-sales records

zero_sales = sales_df[sales_df["Net_Sales_INR"] == 0]

print("Total zero net-sales records:", len(zero_sales))

print("\nReturn reason distribution:")
print(zero_sales["Return_Reason"].value_counts(dropna=False))

print("\nReturned quantity distribution:")
print(zero_sales["Returned_Qty"].value_counts().sort_index())

print("\nRefund amount statistics:")
display(zero_sales["Refund_INR"].describe())
```

```python
# Investigate zero COGS records

zero_cogs = sales_df[sales_df["Net_COGS_INR"] == 0]

print("Total zero COGS records:", len(zero_cogs))

print("\nNet sales distribution:")
print(zero_cogs["Net_Sales_INR"].describe())

print("\nReturned quantity distribution:")
print(zero_cogs["Returned_Qty"].value_counts().sort_index())

print("\nReturn reason distribution:")
print(zero_cogs["Return_Reason"].value_counts(dropna=False))

print("\nZero COGS records with positive net sales:")
print((zero_cogs["Net_Sales_INR"] > 0).sum())
```

```python
# Inspect overall return patterns

print("RETURNED SALES LINES")
print((sales_df["Returned_Qty"] > 0).sum())

print("\nNON-RETURNED SALES LINES")
print((sales_df["Returned_Qty"] == 0).sum())

print("\nTOTAL RETURNED QUANTITY")
print(sales_df["Returned_Qty"].sum())

print("\nRETURN REASON DISTRIBUTION")
print(sales_df["Return_Reason"].value_counts(dropna=False))

print("\nREFUND TOTAL (INR)")
print(round(sales_df["Refund_INR"].sum(), 2))
```

```python
# Validate return reason consistency

returned_reason_mismatch = sales_df[
    (sales_df["Returned_Qty"] > 0) &
    (sales_df["Return_Reason"] == "Not Returned")
]

not_returned_reason_mismatch = sales_df[
    (sales_df["Returned_Qty"] == 0) &
    (sales_df["Return_Reason"] != "Not Returned")
]

print(
    "Returned quantity with Not Returned reason:",
    len(returned_reason_mismatch)
)

print(
    "Zero returned quantity with return reason:",
    len(not_returned_reason_mismatch)
)
```

```python
print("PAYMENT METHOD")
print(sales_df["Payment_Method"].value_counts(dropna=False))

print("\nPROMOTION")
print(sales_df["Promotion"].value_counts(dropna=False))

print("\nSTATE")
print(store_df["State"].value_counts(dropna=False))

print("\nREGION")
print(store_df["Region"].value_counts(dropna=False))

print("\nSPORT")
print(products_df["Sport"].value_counts(dropna=False))

print("\nPRODUCT TYPE")
print(products_df["Product_Type"].value_counts(dropna=False))

print("\nCOLOUR")
print(products_df["Colour"].value_counts(dropna=False))

print("\nBRAND LABEL")
print(products_df["Brand_Label"].value_counts(dropna=False))
```

```python
# Validate product hierarchy consistency

product_type_sport = products_df.groupby(
    "Product_Type"
)["Sport"].nunique()

brand_sport = products_df.groupby(
    "Brand_Label"
)["Sport"].nunique()

print(
    "Product types mapped to multiple sports:",
    (product_type_sport > 1).sum()
)

print(
    "Brand labels mapped to multiple sports:",
    (brand_sport > 1).sum()
)
```

```python
# Check blank values in product and store dimensions

product_columns = [
    "Product_Name", "Sport", "Product_Type",
    "Brand_Label", "Product_Tier", "Colour"
]

store_columns = [
    "Store_Name", "City", "State", "Region",
    "Country", "Store_Format"
]

print("Blank Product Fields:")
print((products_df[product_columns].apply(
    lambda column: column.str.strip().eq("")
)).sum())

print("\nBlank Store Fields:")
print((store_df[store_columns].apply(
    lambda column: column.str.strip().eq("")
)).sum())
```

```python
# Check duplicate product and store descriptions

print(
    "Duplicate Product Names:",
    products_df["Product_Name"].duplicated().sum()
)

print(
    "Duplicate Store Names:",
    store_df["Store_Name"].duplicated().sum()
)

print(
    "Duplicate Store Locations:",
    store_df.duplicated(
        subset=["Store_Name", "City", "State"]
    ).sum()
)
```

```python
# Overall sales and return quantity summary

total_quantity = sales_df["Quantity"].sum()
total_returned_qty = sales_df["Returned_Qty"].sum()

net_quantity = total_quantity - total_returned_qty

return_quantity_rate = (
    total_returned_qty / total_quantity
) * 100

print("Total Quantity Sold:", total_quantity)
print("Total Returned Quantity:", total_returned_qty)
print("Net Quantity After Returns:", net_quantity)
print("Return Quantity Rate:", round(return_quantity_rate, 2), "%")
```

```python
# Validate CY YTD and PYTD date coverage

cy_start = pd.Timestamp("2026-01-01")
cy_cutoff = pd.Timestamp("2026-09-15")

py_start = pd.Timestamp("2025-01-01")
py_cutoff = pd.Timestamp("2025-09-15")

cy_records = sales_df[
    sales_df["Purchase_Date"].between(cy_start, cy_cutoff)
]

py_records = sales_df[
    sales_df["Purchase_Date"].between(py_start, py_cutoff)
]

print("CY YTD Records:", len(cy_records))
print("CY YTD Date Range:",
      cy_records["Purchase_Date"].min(),
      "to",
      cy_records["Purchase_Date"].max())

print("\nPYTD Records:", len(py_records))
print("PYTD Date Range:",
      py_records["Purchase_Date"].min(),
      "to",
      py_records["Purchase_Date"].max())
```

```python
# Identify dates without sales records

expected_dates = pd.date_range(
    start=sales_df["Purchase_Date"].min(),
    end=sales_df["Purchase_Date"].max(),
    freq="D"
)

actual_dates = pd.DatetimeIndex(
    sales_df["Purchase_Date"].dt.normalize().unique()
)

missing_dates = expected_dates.difference(actual_dates)

print("Expected Calendar Days:", len(expected_dates))
print("Dates with Sales:", len(actual_dates))
print("Dates without Sales:", len(missing_dates))

print("\nFirst 20 Dates without Sales:")
print(missing_dates[:20].strftime("%Y-%m-%d").tolist())
```

```python
# Check whether an order has multiple purchase dates

order_date_counts = sales_df.groupby(
    "Order_ID"
)["Purchase_Date"].nunique()

orders_with_multiple_dates = (
    order_date_counts > 1
).sum()

print(
    "Orders with multiple purchase dates:",
    orders_with_multiple_dates
)
```

```python
# Check order-level payment method and promotion consistency

order_payment_counts = sales_df.groupby(
    "Order_ID"
)["Payment_Method"].nunique()

order_promotion_counts = sales_df.groupby(
    "Order_ID"
)["Promotion"].nunique(dropna=False)

print(
    "Orders with multiple payment methods:",
    (order_payment_counts > 1).sum()
)

print(
    "Orders with multiple promotion values:",
    (order_promotion_counts > 1).sum()
)
```

```python
# Summarize financial values at order level

order_summary = sales_df.groupby("Order_ID").agg(
    Order_Net_Sales=("Net_Sales_INR", "sum"),
    Order_Net_COGS=("Net_COGS_INR", "sum"),
    Order_Gross_Profit=("Gross_Profit_INR", "sum")
)

financial_mismatch = (
    order_summary["Order_Gross_Profit"]
    - (
        order_summary["Order_Net_Sales"]
        - order_summary["Order_Net_COGS"]
    )
).abs() > 0.01

print("Total Orders:", len(order_summary))
print(
    "Orders with financial reconciliation mismatch:",
    financial_mismatch.sum()
)
```

```python
# Check duplicate order and product combinations

duplicate_order_product = sales_df.duplicated(
    subset=["Order_ID", "Product_ID"]
).sum()

print(
    "Duplicate Order_ID + Product_ID combinations:",
    duplicate_order_product
)
```

```python
# Validate sales-to-product mapping

sales_product_check = sales_df.merge(
    products_df,
    left_on="Product_ID",
    right_on="Product_Key",
    how="left",
    indicator=True
)

print("Total Sales Lines:", len(sales_product_check))

print(
    "Unmatched Sales Lines:",
    (sales_product_check["_merge"] == "left_only").sum()
)

print(
    "Missing Product Names After Join:",
    sales_product_check["Product_Name"].isna().sum()
)
```

```python
# Validate sales-to-store mapping

sales_store_check = sales_df.merge(
    store_df,
    left_on="Store_ID",
    right_on="Store_Key",
    how="left",
    indicator=True
)

print("Total Sales Lines:", len(sales_store_check))

print(
    "Unmatched Sales Lines:",
    (sales_store_check["_merge"] == "left_only").sum()
)

print(
    "Missing Store Names After Join:",
    sales_store_check["Store_Name"].isna().sum()
)
```

```python
# Check negative financial values

financial_columns = [
    "Unit_List_Price_INR",
    "Unit_Cost_INR",
    "Gross_Sales_INR",
    "Discount_INR",
    "Sales_Before_Returns_INR",
    "Refund_INR",
    "Net_Sales_INR",
    "Net_COGS_INR",
    "Gross_Profit_INR"
]

print("Negative Financial Values:")

print(
    (sales_df[financial_columns] < 0).sum()
)
```

```python
# Validate discount calculations

expected_discount = (
    sales_df["Gross_Sales_INR"]
    * sales_df["Discount_Pct"]
)

discount_difference = (
    sales_df["Discount_INR"] - expected_discount
).abs()

discount_mismatches = (
    discount_difference > 0.01
)

print(
    "Discount Calculation Mismatches:",
    discount_mismatches.sum()
)

print(
    "Maximum Discount Difference:",
    round(discount_difference.max(), 4)
)
```

```python
# Validate refund amounts

refund_exceeds_sales = (
    sales_df["Refund_INR"]
    > sales_df["Sales_Before_Returns_INR"] + 0.01
)

print(
    "Refunds exceeding sales before returns:",
    refund_exceeds_sales.sum()
)
```

```python
# Calculate sales-line gross profit margin

positive_sales = sales_df[
    sales_df["Net_Sales_INR"] > 0
].copy()

positive_sales["Gross_Margin_Pct"] = (
    positive_sales["Gross_Profit_INR"]
    / positive_sales["Net_Sales_INR"]
) * 100

print(
    "Sales Lines with Positive Net Sales:",
    len(positive_sales)
)

print(
    "Minimum Gross Margin (%):",
    round(positive_sales["Gross_Margin_Pct"].min(), 2)
)

print(
    "Maximum Gross Margin (%):",
    round(positive_sales["Gross_Margin_Pct"].max(), 2)
)

print(
    "Negative Gross Margin Lines:",
    (positive_sales["Gross_Margin_Pct"] < 0).sum()
)
```

```python
# Validate product base price and cost

cost_exceeds_price = (
    products_df["Base_Unit_Cost_INR"]
    > products_df["Base_List_Price_INR"]
)

print(
    "Products where base cost exceeds list price:",
    cost_exceeds_price.sum()
)
```

```python
# Compare sales unit list price with product base list price

price_check = sales_df.merge(
    products_df[["Product_Key", "Base_List_Price_INR"]],
    left_on="Product_ID",
    right_on="Product_Key",
    how="left"
)

price_difference = (
    price_check["Unit_List_Price_INR"]
    - price_check["Base_List_Price_INR"]
).abs()

print(
    "Sales lines with different list prices:",
    (price_difference > 0.01).sum()
)

print(
    "Maximum price difference:",
    round(price_difference.max(), 2)
)
```

```python
# Rebuild price comparison with product names

price_check = sales_df.merge(
    products_df[
        ["Product_Key", "Product_Name", "Base_List_Price_INR"]
    ],
    left_on="Product_ID",
    right_on="Product_Key",
    how="left"
)

price_difference = (
    price_check["Unit_List_Price_INR"]
    - price_check["Base_List_Price_INR"]
).abs()

price_mismatches = price_check[
    price_difference > 0.01
]

print("Affected Sales Lines:", len(price_mismatches))

print("\nSample Price Differences:")

display(
    price_mismatches[
        [
            "Purchase_Date",
            "Product_ID",
            "Product_Name",
            "Unit_List_Price_INR",
            "Base_List_Price_INR"
        ]
    ].head(10)
)

print("\nAffected Unique Products:")
print(price_mismatches["Product_ID"].nunique())
```

## 3. Important Business Findings

### 3.1 Sales and Order Structure

- Total sales lines: 84,000.
- Distinct orders: 53,190.
- A single order can contain multiple sales lines.
- Total Orders must be calculated using distinct Order_ID, not sales-line count.

### 3.2 Missing Promotion Values

- Missing Promotion records: 57,131.
- Seasonal Offer: 18,348.
- Weekend Offer: 8,521.

**Decision:** Preserve missing values until their business meaning is confirmed. Do not automatically replace them with "No Promotion".

### 3.3 Returns and Refunds

| Metric | Result |
|---|---:|
| Total quantity sold | 105,444 |
| Total returned quantity | 3,424 |
| Net quantity after returns | 102,020 |
| Returned sales lines | 3,043 |
| Total refunds | ₹4,210,290.03 |
| Fully refunded sales lines | 2,704 |

**Decision:** Fully refunded transactions are valid business records and must not be deleted.

### 3.4 Financial Reconciliation

The following calculations were validated successfully:

1. Gross Sales = Quantity × Unit List Price.
2. Sales Before Returns = Gross Sales − Discount.
3. Net Sales = Sales Before Returns − Refund.
4. Gross Profit = Net Sales − Net COGS.

All reconciliation checks had zero mismatches within the agreed ₹0.01 tolerance.

**Decision:** Use the existing Net Sales, Net COGS and Gross Profit columns. Do not deduct discounts or refunds again.

### 3.5 Product Pricing and Cost Changes

| Year | List Price Change | Unit Cost Change |
|---|---:|---:|
| 2025 | 0% | 0% |
| 2026 | 3% | Approximately 2.5% |

All 38,000 sales lines in 2026 show these systematic changes compared with the product catalogue.

**Decision:** Preserve both transaction values and catalogue values. Record the observed changes without assuming they are officially approved revisions.

### KPI Definitions

- **Net Sales:** SUM(Net_Sales_INR)
- **Gross Profit:** SUM(Gross_Profit_INR)
- **Gross Margin:** Total Gross Profit / Total Net Sales
- **Net Units Sold:** SUM(Quantity − Returned_Qty)
- **Total Orders:** DISTINCTCOUNT(Order_ID)
- **AOV:** Total Net Sales / Distinct Orders
